````python
import pandas as pd, os
ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd())=='notebooks' else os.getcwd()
P = os.path.join(ROOT,'data','processed'); O = os.path.join(ROOT,'outputs')
pv = pd.read_csv(os.path.join(P,'player_values.csv'))
print('Giocatori:', len(pv), '| newcomer:', int(pv.is_newcomer.sum()))
pv.head(3)
````

````python
cols=['Nome','team','role','exp_price','exp_total','VAR','max_bid_static','risk','buyability']
pv.sort_values('exp_total',ascending=False).head(20)[cols]
````
